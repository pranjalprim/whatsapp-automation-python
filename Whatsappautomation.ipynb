```python
# ==========================================
# WhatsApp Web Automation Script (Selenium 4)
# ==========================================

import os  # Import to handle file paths and check file existence
import time  # Import to handle time delays (sleep)
from selenium import webdriver  # Import main WebDriver to control the browser
from selenium.webdriver.common.by import By  # Import element locator strategies (XPATH, CSS, etc.)
from selenium.webdriver.common.keys import Keys  # Import keyboard keys (ENTER, TAB, etc.)
from selenium.webdriver.support.ui import WebDriverWait  # Import explicit waits for elements
from selenium.webdriver.support import expected_conditions as EC  # Import conditions for waiting

# 1. Initialize and launch Chrome browser
driver = webdriver.Chrome()

# 2. Open WhatsApp Web
driver.get("https://web.whatsapp.com")

# 3. Setup dynamic wait (waits up to 600 seconds for QR scan and page load)
wait = WebDriverWait(driver, 600)

# 4. User inputs and variable initialization
target_name = "My"  # Target contact or group name
string = input("Enter the message: ")  # Get text message input from the user
file_path = r"C:\Users\User\Desktop\WhatsApp.jpeg"  # Absolute path to the file/image

# 5. Construct XPath to search for the target chat
x_arg = f'//span[contains(@title, "{target_name}")]'

# 6. Wait until the target chat element appears on the screen
target_element = wait.until(EC.presence_of_element_located((By.XPATH, x_arg)))

# 7. Click on the target chat using JavaScript to avoid click interception
driver.execute_script("arguments[0].click();", target_element)

# 8. Wait for the chat message input field to be clickable
input_box_xpath = '//footer//div[@contenteditable="true"]'
input_box = wait.until(EC.element_to_be_clickable((By.XPATH, input_box_xpath)))

# 9. Focus on the message input box
input_box.click()

# 10. Send text message if provided by the user
if string:
    input_box.send_keys(string + Keys.ENTER)
    time.sleep(1)  # Brief pause to allow text processing

# 11. Handle file attachment and sending logic
if file_path and os.path.exists(file_path):
    
    # Target the hidden <input type="file"> element for media uploads
    media_input_xpath = '//input[@accept="image/*,video/mp4,video/3gpp,video/quicktime"] | //input[@type="file"]'
    file_input = wait.until(EC.presence_of_element_located((By.XPATH, media_input_xpath)))
    
    # Send file path directly to the input element (bypasses Windows file dialog)
    file_input.send_keys(file_path)

    # Pause 2 seconds to allow media preview overlay to load completely
    time.sleep(2)

    # Comprehensive XPath for the green send button container and icon
    send_button_xpath = (
        '//div[@aria-label="Send"]'
        ' | //button[@aria-label="Send"]'
        ' | //span[@data-icon="wds-ic-send-filled"]'
        ' | //span[@data-testid="wds-ic-send-filled"]'
    )
    
    # Wait until the send button element is present in the DOM
    send_button = wait.until(EC.presence_of_element_located((By.XPATH, send_button_xpath)))
    
    # Force click the send button using JavaScript
    driver.execute_script("arguments[0].click();", send_button)
    
    # Wait 3 seconds to ensure media transmission completes
    time.sleep(3)
    print("Image sent successfully!")
else:
    print("File does not exist at the specified path!")

# 12. Wait 5 seconds before closing the browser
time.sleep(5)
# driver.quit()
```
